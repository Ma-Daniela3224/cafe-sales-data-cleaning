# Insights Finales — Analisis de Ventas de Cafeteria

## Resumen ejecutivo

Se analizaron 9,942 transacciones de venta de una cafeteria (dataset sintetico), 
tras un proceso de limpieza que recupero cerca de 500 valores numericos mediante 
relaciones matematicas y estandarizo datos inconsistentes en columnas categoricas.

## Insight 1: Calidad de datos desigual entre columnas

Location (~40%) y Payment Method (~32%) presentan tasas de dato faltante mucho 
mas altas que Item (~10%). Esto sugiere que, en un negocio real, valdria la pena 
revisar el proceso de captura de estos dos campos especificos — posiblemente son 
opcionales o de llenado manual, a diferencia de Item que probablemente se 
selecciona de un catalogo fijo.

## Insight 2: Ventas estables, sin estacionalidad

Las ventas mensuales varian menos del 10% a lo largo del año (entre $6,600 y  
$7,350), sin tendencia de crecimiento ni patron estacional. En un negocio real, 
esto podria interpretarse como una demanda constante, sin eventos o temporadas 
que impacten el consumo.

## Insight 3: Preferencia de producto pareja, sin un "producto estrella"

Los 8 productos tienen volumenes de venta muy similares (diferencia menor al 20% 
entre el mas y menos vendido). No hay evidencia de un producto dominante que 
merezca atencion especial en inventario o promocion.

## Insight 4: Ticket promedio bajo pero consistente

El ticket promedio es de $8.93, con mediana de  $8.0 — muy cercanos, indicando 
consistencia en el gasto por transaccion sin compras atipicas que distorsionen 
el promedio.

## Metodologia de limpieza (resumen)

- Estandarizacion de valores "ERROR"/"UNKNOWN" a NaN
- Recuperacion de ~500 valores numericos via Total Spent = Quantity x Price Per Unit
- Imputacion de categoricas faltantes con etiqueta explicita "Not Specified"
- Eliminacion de ~40 filas (0.4%) sin datos de venta recuperables
- Conservacion de filas sin fecha (4.6%) para analisis no temporales

## Limitaciones

Al ser un dataset sintetico, los patrones (o ausencia de ellos) no deben 
interpretarse como comportamiento real de consumidores. El valor de este 
proyecto es demostrar un flujo de limpieza y analisis riguroso, replicable 
en datos reales.
